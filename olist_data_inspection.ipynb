```python
import numpy as np
import pandas as pd
import zipfile
import os
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
zip_path = r"archive.zip"
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall("dataset")

base = 'dataset'
dfs = {}

for file in os.listdir(base):
    if file.endswith('.csv'):
        dfs[file] = pd.read_csv(os.path.join(base, file))

print("Load Tables:", list(dfs.keys()))
```

```python

oslist_order_payments = pd.read_csv(f'{base}/olist_order_payments_dataset.csv')
product_category_name_translation = pd.read_csv(f'{base}/product_category_name_translation.csv')
olist_sellers = pd.read_csv(f'{base}/olist_sellers_dataset.csv')
olist_customers = pd.read_csv(f'{base}/olist_customers_dataset.csv')
olist_geolocation = pd.read_csv(f'{base}/olist_geolocation_dataset.csv')
olist_products = pd.read_csv(f'{base}/olist_products_dataset.csv')
olist_orders = pd.read_csv(f'{base}/olist_orders_dataset.csv')
olist_order_reviews = pd.read_csv(f'{base}/olist_order_reviews_dataset.csv')
olist_order_items = pd.read_csv(f'{base}/olist_order_items_dataset.csv')
```

```python
tables = {
    'Order_Payment':oslist_order_payments,
    'Category Translation': product_category_name_translation,
    'Sellers': olist_sellers,
    'Customer': olist_customers,
    "Geolocation": olist_geolocation,
    "Products": olist_products,
    "Orders": olist_orders,
    "Order Reviews": olist_order_reviews,
    "Order Items": olist_order_items
}

for name ,df in tables.items():
  print(f"Dataset: {name}")
  print(f'Rows: {df.shape[0]} | Columns: {df.shape[1]}')
  print(f'Duplicates: {df.duplicated().sum()}')

  null_counts = df.isnull().sum()
  null_col = null_counts[null_counts>0]

  if not null_col.empty:
    print('Missing Values:')
    print(null_col.to_string())
  else:
    print('Missing Val not Found')

  print("-" * 40 + "\n")
```
