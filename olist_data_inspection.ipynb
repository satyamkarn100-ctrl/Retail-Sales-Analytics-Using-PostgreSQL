# Imports

```python
import numpy as np
import pandas as pd
import zipfile
import os
```

```python
zip_path = r"D:\Computer science\python lib\project\Retail-Sales-Analytics-Using-PostgreSQL\dataset\archive.zip"

with zipfile.ZipFile(zip_path,'r') as zip_ref:
    zip_ref.extractall("dataset")
base = 'dataset'

dfs = {}
for file in os.listdir(base):
    if file.endswith('.csv'):
        key_name = file.replace('oslist_',"").replace("_dataset.csv",'').replace('.csv','')
        dfs[key_name] = pd.read_csv(os.path.join(base,file))

print("Load Tables:",list(dfs.keys()))
```
